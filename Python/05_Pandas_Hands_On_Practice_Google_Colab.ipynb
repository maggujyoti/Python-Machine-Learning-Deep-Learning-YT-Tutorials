## 7. Creating new columns

Let's calculate a total score and an average score.


```python
df["Total"] = df["Math"] + df["AI_Score"]
df["Average"] = df["Total"] / 2

df[["Name", "Math", "AI_Score", "Total", "Average"]].head()
```

### Create a result column

A student passes if Average >= 60.


```python
df["Result"] = np.where(df["Average"] >= 60, "Pass", "Fail")
df[["Name", "Average", "Result"]].head(10)
```

## 8. Missing values

Notice that some Math and AI_Score values are missing.


```python
print(df.isna().sum())
```

```python
missing_rows = df[df.isna().any(axis=1)]
missing_rows
```

### Fill missing numeric values

For this demonstration, use the column mean.


```python
df["Math"] = df["Math"].fillna(df["Math"].mean())
df["AI_Score"] = df["AI_Score"].fillna(df["AI_Score"].mean())

print(df.isna().sum())
```

## 9. GroupBy

Let's calculate average Math and AI scores by department.


```python
department_avg = df.groupby("Department")[["Math", "AI_Score"]].mean()
department_avg
```

### Group by city


```python
city_stats = df.groupby("City")["Average"].agg(["mean", "max", "min", "count"])
city_stats.sort_values("mean", ascending=False)
```

## 16. Final one-cell workflow

This is the pattern to remember:

**Load -> Inspect -> Clean -> Transform -> Analyze -> Save**


```python
df = pd.read_csv("pandas_students.csv")

df["Math"] = df["Math"].fillna(df["Math"].mean())
df["AI_Score"] = df["AI_Score"].fillna(df["AI_Score"].mean())

df["Total"] = df["Math"] + df["AI_Score"]
df["Average"] = df["Total"] / 2

result = (
    df.groupby("Department")["Average"]
      .mean()
      .sort_values(ascending=False)
)

print(result)

df.to_csv("student_analysis.csv", index=False)
```
