# Functions in Python


```python
def greet():
    print('Hello Students!')

greet()
```

```python
def greet(name):
    print('Hello', name)

greet('Alice')
greet('Bob')
```

```python
def add(a,b):
    return a+b

result=add(10,20)
print(result)
```

```python
print(len([1,2,3]))
print(max([10,4,15]))
print(sum([2,4,6]))
```

```python
def welcome(name='Student'):
    print('Welcome',name)

welcome()
welcome('Rahul')
```

```python
def student(name,age):
    print(name, age)

student(age=20,name='Priya')
```

```python
def total(*numbers):
    return sum(numbers)

print(total(1,2,3,4,5))
```

```python
def details(**info):
    for k,v in info.items():
        print(k,':',v)

details(name='Riya',city='Delhi')
```

```python
square=lambda x:x*x
print(square(9))
```

```python
nums=[1,2,3,4,5]
print(list(map(lambda x:x*x,nums)))
```

```python
nums=[10,11,12,13,14]
print(list(filter(lambda x:x%2==0,nums)))
```

```python
def factorial(n):
    if n==1:
        return 1
    return n*factorial(n-1)

print(factorial(5))
```

```python
x=100

def demo():
    y=50
    print('Global',x)
    print('Local',y)

demo()
```
