## 1. Import Matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np

print("Matplotlib is ready!")
```

## 2. Your First Line Plot
A line plot is useful when order or continuity matters.

```python
x = [1, 2, 3, 4, 5]
y = [10, 15, 13, 20, 25]

plt.plot(x, y)
plt.show()
```

### Add title and axis labels

```python
plt.plot(x, y)

plt.title("Simple Line Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.show()
```

## 3. Line Plot — Monthly Sales
This matches the sales example used in the slides.

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [42, 48, 45, 57, 63, 72]

plt.plot(months, sales)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

### Style the line

```python
plt.plot(
    months,
    sales,
    marker="o",
    linewidth=2
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(alpha=0.25)

plt.show()
```

## 4. Bar Chart — Average Score by Subject

```python
subjects = ["Python", "NumPy", "Pandas", "AI"]
scores = [78, 86, 82, 91]

plt.bar(subjects, scores)

plt.title("Average Score by Subject")
plt.xlabel("Subject")
plt.ylabel("Score")
plt.ylim(0, 100)

plt.show()
```

### Improve the bar chart

```python
plt.bar(subjects, scores)

plt.title("Average Score by Subject")
plt.xlabel("Subject")
plt.ylabel("Score")
plt.ylim(0, 100)
plt.grid(axis="y", alpha=0.2)

plt.show()
```

## 5. Scatter Plot — Study Hours vs Exam Score
A scatter plot helps investigate relationships between two numerical variables.

```python
hours = [1, 2, 2.5, 3, 4, 4.5, 5, 6, 7, 8]
marks = [48, 54, 58, 61, 67, 70, 73, 80, 84, 91]

plt.scatter(hours, marks)

plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.grid(alpha=0.2)

plt.show()
```

### Think about the pattern
The points generally move upward. This suggests a positive relationship in this small example, but a visual pattern alone does not prove causation.

```python
plt.scatter(hours, marks, s=55)

plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")

plt.show()
```

## 6. Histogram — Distribution of Exam Scores
A histogram groups numerical values into **intervals called bins**. Each bar represents the number of observations falling within an interval.

```python
exam_scores = [
    45, 48, 52, 55, 57, 58, 60, 61, 62, 64,
    65, 66, 67, 68, 69, 70, 70, 71, 72, 72,
    73, 74, 74, 75, 76, 77, 78, 79, 80, 80,
    81, 82, 83, 84, 85, 86, 87, 88, 89, 90,
    91, 92, 93, 94, 95
]

plt.hist(
    exam_scores,
    bins=range(40, 101, 10),
    edgecolor="black"
)

plt.title("Distribution of Exam Scores")
plt.xlabel("Score")
plt.ylabel("Number of Students")
plt.grid(axis="y", alpha=0.2)

plt.show()
```

### Understanding bins

```python
# Try fewer bins
plt.hist(exam_scores, bins=5, edgecolor="black")
plt.title("Exam Scores — 5 Bins")
plt.xlabel("Score")
plt.ylabel("Number of Students")
plt.show()
```

```python
# Try more bins
plt.hist(exam_scores, bins=14, edgecolor="black")
plt.title("Exam Scores — 14 Bins")
plt.xlabel("Score")
plt.ylabel("Number of Students")
plt.show()
```

## 7. Multiple Series + Legend

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

python_students = [35, 42, 48, 54, 61, 67]
pandas_students = [28, 34, 41, 45, 52, 59]

plt.plot(
    months,
    python_students,
    marker="o",
    label="Python"
)

plt.plot(
    months,
    pandas_students,
    marker="o",
    label="Pandas"
)

plt.title("Course Progress")
plt.xlabel("Month")
plt.ylabel("Students")
plt.legend()
plt.grid(alpha=0.2)

plt.show()
```

## 8. Subplots
A single Figure can contain several related charts.

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 7))

# Line plot
axes[0, 0].plot(months, sales, marker="o")
axes[0, 0].set_title("Trend")

# Bar chart
axes[0, 1].bar(subjects, scores)
axes[0, 1].set_title("Comparison")
axes[0, 1].tick_params(axis="x", rotation=25)

# Scatter plot
axes[1, 0].scatter(hours, marks)
axes[1, 0].set_title("Relationship")

# Histogram
axes[1, 1].hist(exam_scores, bins=8)
axes[1, 1].set_title("Distribution")

fig.tight_layout()
plt.show()
```

## 9. Figure Size and Layout

```python
plt.figure(figsize=(10, 5))

plt.plot(months, sales, marker="o")
plt.title("Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.tight_layout()
plt.show()
```

## 10. Annotations
Use annotations to highlight an important point.

```python
plt.figure(figsize=(8, 5))

plt.plot(months, sales, marker="o")

plt.annotate(
    "Peak",
    xy=("Jun", 72),
    xytext=("Apr", 68),
    arrowprops={"arrowstyle": "->"}
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

## 11. Saving a Chart
Save the visualization before displaying it.

```python
plt.figure(figsize=(8, 5))

plt.plot(months, sales, marker="o")
plt.title("Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.savefig(
    "sales_trend.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
